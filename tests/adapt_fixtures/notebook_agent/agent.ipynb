# Bookbot

A tiny scheduling assistant for the Ledger team, written in a notebook.

Usage example:

    ask("What is on the calendar for Friday?")

It searches the team calendar and reports the matching event.

```python
import anthropic

client = anthropic.Anthropic()

SYSTEM_PROMPT = (
    "You are Bookbot, a scheduling assistant for the Ledger team. "
    "Search the calendar before you answer, and never invent an event."
)
```

```python
TOOLS = [
    {
        "name": "search_calendar",
        "description": "Search the team calendar for matching events.",
        "input_schema": {
            "type": "object",
            "properties": {
                "query": {"type": "string", "description": "What to search for."}
            },
            "required": ["query"],
        },
    }
]
```

```python
def ask(question):
    return client.messages.create(
        model="claude-fable-5",
        max_tokens=1024,
        system=SYSTEM_PROMPT,
        tools=TOOLS,
        tool_choice={"type": "any"},
        messages=[{"role": "user", "content": question}],
    )
```
